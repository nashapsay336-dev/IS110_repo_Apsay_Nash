```python
total = 0
count = 0

while True:
    user_input = input("Enter a number: ")

    if user_input == "done":
        break

    try:
        num = float(user_input)
        total += num
        count += 1
    except:
        print("Invalid input")
        continue

if count > 0:
    average = total / count
else:
    average = 0

print(total, count, average)
```

```python
largest = None
smallest = None

while True:
    user_input = input("Enter a number: ")

    if user_input == "done":
        break

    try:
        num = float(user_input)
    except:
        print("Invalid input")
        continue

    if largest is None or num > largest:
        largest = num
    if smallest is None or num < smallest:
        smallest = num

print("Maximum:", largest)
print("Minimum:", smallest)
```
